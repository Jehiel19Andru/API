# Maquina De Soporte Vectorial

[SVM](https://unb.ca/cic/datasets/url-2016.html) Una máquina de soporte vectorial (SVM) es un **algoritmo de aprendizaje automático que clasifica datos y resuelve problemas de regresión**. Se trata de una herramienta que se utiliza en muchas áreas, como la biología, las finanzas, el marketing, el reconocimiento de imágenes, el procesamiento de lenguaje natural y la detección de fraudes. 

* Las SVM se basan en la idea de encontrar un hiperplano que separe de manera óptima las diferentes clases de datos. Para ello, se maximiza la distancia entre los puntos más cercanos de cada clase, llamados vectores de soporte. 
* Las SVM son especialmente útiles para conjuntos de datos pequeños con pocos valores atípicos. Se consideran una metodología flexible y con buen rendimiento en un amplio abanico de situaciones.
